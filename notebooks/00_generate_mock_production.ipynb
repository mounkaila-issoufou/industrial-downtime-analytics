```python
import pandas as pd
import random
import uuid
from datetime import datetime, timedelta
```

```python
# =====================================================
# CONFIG
# =====================================================

FACTORY_ID = "F_DOM"
WORKSHOP_ID = "W_OV"
LINE_ID = "L_OV_MS"
MACHINE_ID = "M_MS"
OPERATOR_ID = "OP_014"

CAPACITY_PER_HOUR = 4800

SESSIONS = {
    "MATIN": list(range(5, 13)),      # 5h → 12h
    "SOIR":  list(range(13, 21)),     # 13h → 20h
    "NUIT":  [21, 22, 23, 0, 1, 2, 3, 4],
    "SD":    [6, 7, 8, 9, 7, 10, 11, 12, 13, 14]
}

ORGANS_ELEMENTS = {
    "Empileur": [
        "Défaut porte", "Trainard entrée", "Trainard sortie",
        "Pied de base", "Claie coincée", "Delta 1", "Gavage claie vide"
    ],
    "Emballeuse": [
        "Erreur cadençage", "Défaut stickers",
        "Reprise de film", "Bourrage convoyeur sortie"
    ]
}

ACTIONS = [
    "Réarmement", "Dégagement manuel",
    "Réglage mécanique", "Nettoyage", "Reset automate"
]
```

```python
# =====================================================
# UTILS
# =====================================================

def gen_id(prefix: str) -> str:
    return f"{prefix}_{uuid.uuid4().hex[:8]}"

def rotate_session(day_index: int) -> str:
    return ["MATIN", "SOIR", "NUIT"][day_index % 3]
```

```python
hourly_rows = []
event_rows = []

for day in range(7):
    current_date = start + timedelta(days=day)
    weekday = current_date.weekday()

    session = "SD" if weekday >= 5 else rotate_session(day)
    hours = SESSIONS[session]

    for hour in hours:
        prod_id = gen_id("HP")
        non_prod_minutes = 0

        # événements détaillés
        nb_events = random.randint(1, 4)
        print(f"Prod ID: {prod_id} ===================================")
        for _ in range(nb_events):
            organ = random.choice(list(ORGANS_ELEMENTS.keys()))
            element = random.choice(ORGANS_ELEMENTS[organ])
            duration = random.randint(3, 60)

            non_prod_minutes += duration
            print(f"  prod_id: {prod_id}")
            event_rows.append({
                "event_id": gen_id("EV"),
                "prod_id": prod_id,
                "date": current_date.date(),
                "hour": hour,
                "factory_id": FACTORY_ID,
                "workshop_id": WORKSHOP_ID,
                "line_id": LINE_ID,
                "machine_id": MACHINE_ID,
                "operator_id": OPERATOR_ID,
                "session": session,
                "organ": organ,
                "element": element,
                "duration_minutes": duration,
                "action": random.choice(ACTIONS),
                "comment": "Défaut détecté et traité par l’opérateur"
            })

        non_prod_minutes = min(non_prod_minutes, 55)

        actual_production = int(
            CAPACITY_PER_HOUR * (1 - non_prod_minutes / 60)
        )

        hourly_rows.append({
            "prod_id": prod_id,
            "date": current_date.date(),
            "hour": hour,
            "factory_id": FACTORY_ID,
            "workshop_id": WORKSHOP_ID,
            "line_id": LINE_ID,
            "machine_id": MACHINE_ID,
            "operator_id": OPERATOR_ID,
            "session": session,
            "theoretical_capacity": CAPACITY_PER_HOUR,
            "actual_production": actual_production,
            "non_production_minutes": non_prod_minutes
        })
```

```python
def generate_week(start_date: str):
    hourly_rows = []
    event_rows = []

    start = datetime.strptime(start_date, "%Y-%m-%d")

    for day in range(7):
        current_date = start + timedelta(days=day)
        weekday = current_date.weekday()

        session = "SD" if weekday >= 5 else rotate_session(day)
        hours = SESSIONS[session]

        for hour in hours:
            prod_id = gen_id("HP")
            non_prod_minutes = 0

            # événements détaillés
            nb_events = random.randint(1, 4)

            for _ in range(nb_events):
                organ = random.choice(list(ORGANS_ELEMENTS.keys()))
                element = random.choice(ORGANS_ELEMENTS[organ])
                duration = random.randint(3, 12)

                non_prod_minutes += duration

                event_rows.append({
                    "event_id": gen_id("EV"),
                    "prod_id": prod_id,
                    "date": current_date.date(),
                    "hour": hour,
                    "factory_id": FACTORY_ID,
                    "workshop_id": WORKSHOP_ID,
                    "line_id": LINE_ID,
                    "machine_id": MACHINE_ID,
                    "operator_id": OPERATOR_ID,
                    "session": session,
                    "organ": organ,
                    "element": element,
                    "duration_minutes": duration,
                    "action": random.choice(ACTIONS),
                    "comment": "Défaut détecté et traité par l’opérateur"
                })

            non_prod_minutes = min(non_prod_minutes, 55)

            actual_production = int(
                CAPACITY_PER_HOUR * (1 - non_prod_minutes / 60)
            )

            hourly_rows.append({
                "prod_id": prod_id,
                "date": current_date.date(),
                "hour": hour,
                "factory_id": FACTORY_ID,
                "workshop_id": WORKSHOP_ID,
                "line_id": LINE_ID,
                "machine_id": MACHINE_ID,
                "operator_id": OPERATOR_ID,
                "session": session,
                "theoretical_capacity": CAPACITY_PER_HOUR,
                "actual_production": actual_production,
                "non_production_minutes": non_prod_minutes
            })

    return pd.DataFrame(hourly_rows), pd.DataFrame(event_rows)

# =====================================================
# RUN
# =====================================================

if __name__ == "__main__":

    df_hourly, df_events = generate_week("2024-06-03")

    df_hourly.to_csv("hourly_production.csv", index=False)
    df_events.to_csv("production_events.csv", index=False)

    print("✅ Modèle horaire propre + événements détaillés générés")
```
